## Findings and decisions
- Stripped a stray space from a column name. Dropped the row counter `so`, plus `date` and `zipcode`. Kept `id` for a grouped split.
- `id` repeats 177 times (houses sold more than once): rows kept, the split must be grouped by id.
- Bedrooms = 33 is impossible (1,620 sqft): set to missing. Bedrooms = 11 kept.
- `yr_renovated` was 0 for 96% of rows (a code for "never renovated"): replaced by `renovated` (0/1).
- Remaining missing values (bedrooms, bathrooms) are filled inside the pipeline.
- Output: data/processed/clean.csv (id, price and 17 features).
